```python
## Q2 Count Commas in Range II

class Solution:
    def countCommas(self, n: int) -> int:
        total = 0
        comma_count = 1
        lower = 1000
        
        while lower <= n:
            upper = min(n, 10**((comma_count+1)*3) - 1)
            total += (upper - lower + 1) * comma_count
            lower *= 1000
            comma_count += 1
            
        return total


s = Solution()
n = 1004590
print(s.countCommas(n))


# Input: n = 1002
# Output: 3

# Input: n = 998
# Output: 0

# Input: n = 1004590
# Output: 1008182
```

```python
## Q3 Longest Arithmetic Sequence After Changing At Most One Element©leetcode

from typing import List
class Solution:
    def longestArithmetic(self, nums: List[int]) -> int:
        d={}
        for i in range(1,len(nums)):
            d[i]=abs(nums[i]-nums[i-1])
```
