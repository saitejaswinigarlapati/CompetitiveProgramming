```python
# Two Sum

## Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.Array index starts from 1.

def twosum(arr,target):
    left=0
    right=len(arr)-1
    while(left<right):
        if arr[left]+arr[right] == target:
            return [left+1,right+1]
        elif arr[left]+arr[right] < target:
            left+=1
        else:
            right-=1
    return -1

arr=[2,7,11,15]
target=9

print(twosum(arr,target))
```

```python
def pairsum(arr,k):
    l=0
    r=1
    c=0
    while r<len(arr):
        s=arr[r]-arr[l]
        if s==k:
            c+=1
            l+=1
            r+=1
        elif s>k:
            l+=1
        else:
            r+=1
        if l==r:
            r+=1
    return c

print(pairsum([1,2,3,4],1))
```

```python
# Leetcode 167 : Two Sum 2

from typing import List
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        n=len(numbers)
        left=0
        right=n-1
        while left < right:
            sum=numbers[left]+numbers[right]
            if sum == target:
                return [left+1,right+1]
            elif sum > target:
                right-=1
            else:
                left+=1
        
s=Solution()
numbers=[2,7,11,15]
target=9
print(s.twoSum(numbers,target))

# Input: numbers = [2,3,4], target = 6
# Output: [1,3]

# Input: numbers = [-1,0], target = -1
# Output: [1,2]
```

```python
# Leetcode 75: Sort Colors

class Solution:
    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n=len(nums)
        red=0
        white=0
        blue=n-1
        while white <= blue:
            if nums[white]==2:
                nums[white],nums[blue]=nums[blue],nums[white]
                blue-=1
            elif nums[white]==0:
                nums[white],nums[red]=nums[red],nums[white]
                red+=1
                white+=1
            else:
                white+=1
        return nums
        
s=Solution()
nums=[2,0,2,1,1,0]
print(s.sortColors(nums))
```

```python
# Leetcode 15: 3Sum
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        n=len(nums)
        if n < 3: return []
        result=set()
        nums.sort()
        for i in range(n-2):
            left=i+1
            right=n-1
            while left<right:
                total=nums[left]+nums[right]+nums[i]
                if total==0:
                    result.add((nums[i],nums[left],nums[right]))
                    left+=1
                    right-=1
                elif total <0:
                    left+=1
                else:
                    right-=1
        return [list(triplet) for triplet in result]
    
s=Solution()
nums=[-1,0,1,2,-1,-4]
print(s.threeSum(nums))
```

```python
# Leetcode 283: Move Zeroes

class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        
        left=0
        right=0
        while right <len(nums):
            if nums[right]!=0:
                nums[left],nums[right]=nums[right],nums[left]
                left+=1
                right+=1
            else:
                right+=1
        
s=Solution()
nums=[0,1,0,3,12]
s.moveZeroes(nums)
print(nums)
```

```python
# Container With Most Water

class Solution:
    def containerwithmostwater(self,height: List[int]):
        left=0
        right=len(height)-1
        max_sum=0
        while left<=right:
            w=right-left
            h=min(height[left],height[right])
            max_sum=max(max_sum,w*h)
            
            if height[left]<height[right]:
                left+=1
            else:
                right-=1
        return max_sum
    
s=Solution()
height=[1,8,6,2,5,4,8,3,7]
print(s.containerwithmostwater(height))
```

```python
# Leetcode 42: Trapping rain water

class Solution:
    def trap(self, height: List[int]) -> int:
        n=len(height)
        left=0
        right=n-1
        l_m=height[left]
        r_m=height[right]
        water=0
        while left<right:
            if height[left]<height[right]:
                left+=1
                l_m=max(l_m,height[left])
                water+=l_m-height[left]
            else:
                right-=1
                r_m=max(r_m,height[right])
                water+=r_m-height[right]
        return water
    
s=Solution()
height=[0,1,0,2,1,0,1,3,2,1,2,1]
print(s.trap(height))
```
