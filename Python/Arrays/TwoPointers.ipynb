```python
# Two Sum

## Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.Array index starts from 1.

def twosum(arr,target):
    left=0
    right=len(arr)-1
    while(left<right):
        if arr[left]+arr[right] == target:
            return [left+1,right+1]
        elif arr[left]+arr[right] < target:
            left+=1
        else:
            right-=1
    return -1

arr=[2,7,11,15]
target=9

print(twosum(arr,target))
```
