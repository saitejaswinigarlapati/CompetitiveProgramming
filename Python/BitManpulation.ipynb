```python
## Find the missing number in the arr with n numbers

def find_missing(arr,n):
    xor_full=0
    xor_arr=0
    
    for i in range(n+1):
        xor_full^=i
    
    for num in arr:
        xor_arr^=num
    
    missing=xor_full ^ xor_arr
    return missing

arr=[0,1,2,3,4,5,7,8]
n=8
print("Missing number: ",find_missing(arr,n))
```

```python
## Toggle kth bit in the number

def toggle_kth_bit(n,k):
    a=1<<k
    return (n ^ a)

n=16
k=2
print(toggle_kth_bit(n,k))
```

```python
# Check whether a number is even or odd with out mod operator

def even_or_odd(num):
    s=num & 1
    if (s==1):
        return "Odd"
    return "Even"
even_or_odd(3)
```

```python
# Swap to numbers with out temp variable

def swap(a,b):
    a= a^b
    b=a^b
    a=a^b
    return (a,b)

a=2
b=3
swap(a,b)
```

```python
# Number of 1 bits in the given number
def count(n):
    c=0
    while n!=0:
        n=n & n-1
        c+=1
    return c
count(11)
```

```python
# Reverse Bits
def reverseBits(n: int) -> int:
    res = 0
    for i in range(32):
        res = (res << 1) | (n & 1)
        n >>= 1
    return res

n = 2147483644
reverseBits(n)
```
