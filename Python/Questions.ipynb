1. A gym generates a daily door lock code for members.

    * The lock code changes based on the day (1 = Monday … 7 = Sunday).

    * To generate it: take the first 3 letters of the day, reverse them, make them uppercase.

```python
def generatekey(dayNumber):
    days={1:"Monday",2:"Tuesday",3:"Wednesday",4:"Thursday",5:"Friday",6:"Saturday",7:"Sunday"}
    if dayNumber<0 or dayNumber > 7:
        return "Invalid day number "
    dayName=days[dayNumber]
    key=dayName[:3][::-1].upper()
    return key

day=int(input("Enter day number: "))
print("Day Number : ",day)
print("Key : ",generatekey(day))
    
print("Day Number : ",2)
print("Key : ",generatekey(2))
    
```

2. A bank stores the daily transactions (in rupees) of its customers in an array. Each element in the array represents the amount spent in a transaction. The bank wants to:

    * Find the highest transaction amount made on that day.
    * Calculate the sum of digits of that amount (for example, for ₹753 → 7+5+3 = 15).
    * This digit sum will be used as a verification code in their system.
    
**Task:**  Write a program that takes the transaction amounts as input, finds the highest transaction, and prints both the largest transaction and its digit sum (verification code).


```python
def transaction(transactionAmount):
    highestTransaction=max(transactionAmount)
    digitSum=0
    for i in str(highestTransaction):
        digitSum+=int(i)
    print("Highest Transaction Amount: ",highestTransaction)
    print("Digit Sum: ",digitSum)
    
    
daily_transactions = [1500, 200, 753, 900, 120]
transaction(daily_transactions)
```

3. Imagine a cricket tournament. Each player has a jersey number from 1 to n:

* If a jersey number is divisible by 2 and 9, the player is marked as “Star Player”.
* If the number is not “Star Player” but divisible by 2, the player is in “Team Blue”.
* Otherwise, the player is in “Team Red”.

**Task:** Write a program that categorizes players based on their jersey numbers.

```python
def categorizePlayers(jerseyNumber):
    if jerseyNumber%2 ==0 and jerseyNumber %9 ==0:
        print("Jersey Number ",jerseyNumber, ": Start Player")
    elif jerseyNumber%2==0:
        print("Jersey Number ",jerseyNumber,": Team Blue")
    else:
        print("Jersey Number ",jerseyNumber,": Team Red")
        
jerseyNumber1=9
categorizePlayers(jerseyNumber1)

jerseyNumber2=12
categorizePlayers(jerseyNumber2)

jerseyNumber3=18
categorizePlayers(jerseyNumber3)
```

A company’s door lock uses a special code system.

The valid codes are Fibonacci numbers that are also prime, but they cannot exceed a maximum value n.

**Task:** Given n, print all the valid door codes.

```python
def isPrime(num):
    if num<=1:
        return False
    for i in range(2,int(num**0.5)+1):
        if num%i == 0:
            return False
    return True

def validDoorNumbers(n):
    fib1, fib2 = 0, 1
    while fib1 <= n:
        if isPrime(fib1):
            print(fib1, end=" ")
        fib1, fib2 = fib2, fib1 + fib2

n=500
validDoorNumbers(n)
```

 A soldier wants to buy w bananas in the shop. He has to pay k dollars for the first banana, 2k dollars for the second one and so on (in other words, he has to pay i·k dollars for the i-th banana). He has n dollars. How many dollars does he have to borrow from his friend soldier to buy w bananas?

Input:

The first line contains three positive integers k, n, w (1 ≤ k, w ≤ 1000, 0 ≤ n ≤ 10⁹), the cost of the first banana, the initial number of dollars the soldier has, and the number of bananas he wants.

Output:

Output one integer — the amount of dollars that the soldier must borrow from his friend. If he doesn’t have to borrow money, output 0.

Input:
3 17 4

Output:
13

```python
def buyBananas(k,n,w):
    c=0
    for i in range(1,w+1):
        c+=i*k
    return c-n

print("Amount of dollars that the soldier must borrow from his friend ",buyBananas(3,17,4))
```

One day three best friends Petya, Vasya and Tonya decided to form a team and take part in programming contests. Participants are usually offered several problems during programming contests. Long before the start the friends decided that they will implement a problem if at least two of them are sure about the solution. Otherwise, the friends won't write the problem's solution. This contest offers n problems to the participants. For each problem we know, which friend is sure about the solution. Help the friends find the number of problems for which they will write a solution.

Input

The first input line contains a single integer n (1 ≤ n ≤ 1000) — the number of problems in the contest. Then n lines contain three integers each, each integer is either 0 or 1. If the first number in the line equals 1, then Petya is sure about the problem's solution, otherwise he isn't sure. The second number shows Vasya's view on the solution, the third number shows Tonya's view. The numbers on the lines are separated by spaces.

Output

Print a single integer — the number of problems the friends will implement on the contest.

Input

3
1 1 0
1 1 1
1 0 0

Output

2

```python
def numberOfProblems(n):
    count=0
    for _ in range(n):
        a,b,c=map(int,input().split(" "))
        if (a+b+c>=2):
            count+=1
    return count

s=numberOfProblems(3)
print("Number of problems the friends will implement on the contest : ",s)
```
