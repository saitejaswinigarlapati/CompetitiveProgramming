```python
class Solution:
    def minimumIndex(self, capacity: list[int], itemSize: int) -> int:
        res, idx = float('inf'), -1
        for i, cap in enumerate(capacity):
            if itemSize <= cap < res:
                res = cap
                idx = i
        return idx

s = Solution()
capacity = [1, 5, 3, 7]
itemSize = 3

print(s.minimumIndex(capacity, itemSize))
```

```python
class Solution:
    def smallestBalancedIndex(self, nums: list[int]) -> int:
        n, left, rght = len(nums), sum(nums), 1
        for i in range(n - 1, -1, -1):
            left -= nums[i]
            if   left > rght:
                rght *= nums[i]
            elif left < rght:
                break
            else:
                return i
        return -1
    
s=Solution()
nums = [2,1,2]
print(s.smallestBalancedIndex(nums))
```
