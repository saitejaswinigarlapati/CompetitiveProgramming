**Q1**

```python
def min_swaps_to_sort(heights):
    n = len(heights)
    # Pair height with original index
    arr = [(heights[i], i) for i in range(n)]
    
    # Sort by height, stable sort preserves order for duplicates
    arr.sort()
    
    visited = [False] * n
    swaps = 0
    
    for i in range(n):
        if visited[i] or arr[i][1] == i:
            continue
        
        cycle_size = 0
        j = i
        while not visited[j]:
            visited[j] = True
            j = arr[j][1]
            cycle_size += 1
        
        if cycle_size > 1:
            swaps += cycle_size - 1
    
    return swaps


# Example usage
heights = [2, 4, 7, 7, 9, 3]
print("Minimum number of swaps required: ",min_swaps_to_sort(heights))
```

```python
def max_applicants_with_apartments(n, m, k, applicants, apartments):
    applicants.sort()
    apartments.sort()
    
    i = j = matches = 0
    
    while i < n and j < m:
        if apartments[j] < applicants[i] - k:
            j += 1
        elif apartments[j] > applicants[i] + k:
            i += 1
        else:
            # Match found
            matches += 1
            i += 1
            j += 1
    
    return matches


# Example usage
n, m, k = 4, 3, 5
applicants = [60, 45, 80, 60]
apartments = [30, 60, 75]

print("Number of applicants who will get an apartment: ",max_applicants_with_apartments(n, m, k, applicants, apartments))
```

```python
def min_gondolas(n, x, weights):
    weights.sort()
    i, j = 0, n - 1
    gondolas = 0
    
    while i <= j:
        if weights[i] + weights[j] <= x:
            i += 1  # lightest child goes with heaviest
        j -= 1  # heaviest child always goes
        gondolas += 1
    
    return gondolas


# Example usage
n, x = 4, 10
weights = [7, 2, 3, 9]

print("The minimum number of gondolas: ",min_gondolas(n, x, weights))
```
