```python
## Q1 .Trim Trailing Vowels
##              -- Leetocde question number: 3856

class Solution:
    def trimTrailingVowels(self,s:str) -> str:
        v={'a','e','i','o','u'}
        n=len(s)-1
        while n >=0 and s[n] in v:
            n=n-1
        return s[:n+1]
    
obj=Solution()
s='idea'
print(obj.trimTrailingVowels(s))
```

```python
## Q2. Minimum Cost to Split into Ones
##              -- Leetcode question number : 3875

class Solution:
    def minCost(self, n: int) -> int:
        cost=0
        while n>1:
            a=1
            b=n-1
            cost+=a*b
            n=n-1
        return cost
    
obj=Solution()
print(obj.minCost(3))
            
```

```python
## Optinized Version

class Solution:
    def minCost(self, n: int) -> int:
        return n*(n-1) //2
    
obj=Solution()
print(obj.minCost(3))
```

```python
## Q3. Minimum Bitwise OR From Grid
##              -- Leetcode question number: 3858

from typing import List

## Brute Force Approch - Time Limit Exceedes for large input.
class Solution:
    def minimumOR(self, grid: List[List[int]]) -> int:
        possible ={0}
        for row in grid:
            new_possible=set()
            for prev in possible:
                for num in row:
                    new_possible.add(prev|num)
            possible=new_possible
        return min(possible)
    
obj=Solution()
grid = [[1,5],[2,4]]

print(obj.minimumOR(grid))
```
