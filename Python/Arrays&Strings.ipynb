```python
## Move All Zeros to end

def moveAllzeroestoEnd(ls):
    l=0
    r=len(ls)-1
    print(l,r)
    while l<=r:
        if ls[l]==0:
            ls[l],ls[r]=ls[r],ls[l]
            r-=1
        else:
            l+=1
    return ls

l=[0,2,3,0,5,4,0]
print(moveAllzeroestoEnd(l))
```

```python
## Reverse an Array or String

def reverseArrayorString(obj):
    return obj[::-1]

s='string'
a=[1,2,3,4]
print(reverseArrayorString(s))
print(reverseArrayorString(a))
```

```python
## Check if a string is palindrome or not

def checkPalindrome(str):
    str=str.replace(" ","").lower()
    return str == str[::-1]

s="abcbac"
print(checkPalindrome(s))
```

```python
## Find the missing number in array 1..N

def findMissing(arr):
    s=sum(arr)
    arr.sort()
    n=arr[-1]
    total=n*(n+1)//2
    return total-s

ls=[1,2,3,5,6,7,4,9]
print(findMissing(ls))
```
