#### Search in a Rotated Sorted Array
You are given an array nums sorted in ascending order, rotated at some pivot unknown to you beforehand (e.g., [4,5,6,7,0,1,2]).

Write a function to search for a target value in nums. If the target exists, return its index. Otherwise, return -1.

```python
from typing import List

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                return mid
            
            # Left half is sorted
            if nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            # Right half is sorted
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        
        return -1

nums=[4,5,6,7,0,1,2]
target=0
s=Solution()
print(s.search(nums,target))
```
