Given weights of n children and a gondola capacity C, find the minimum number of gondolas needed.

Each gondola can carry at most two children, and the sum of their weights must be ≤ C.


Input:
weights = [40, 50, 60, 70]
capacity = 100

Output:
3

Explanation:
Gonlands used:
(40 + 60) → 100
(50) → 50
(70) → 70


```python
def min_gonlands(weights, capacity):
    weights.sort()
    left, right = 0, len(weights) - 1
    gondolas = 0

    
    while left <= right:
        if weights[left] + weights[right] <= capacity:
            left += 1
        right -= 1
        gondolas += 1

    return gondolas

print(min_gonlands([40, 50, 60, 70], 100))  # Output: 3
```
