Given weights of n children and a gondola capacity C, find the minimum number of gondolas needed.

Each gondola can carry at most two children, and the sum of their weights must be ≤ C.


Input:
weights = [40, 50, 60, 70]
capacity = 100

Output:
3

Explanation:
Gonlands used:
(40 + 60) → 100
(50) → 50
(70) → 70


```python
def min_gonlands(weights, capacity):
    weights.sort()
    left, right = 0, len(weights) - 1
    gondolas = 0

    
    while left <= right:
        if weights[left] + weights[right] <= capacity:
            left += 1
        right -= 1
        gondolas += 1

    return gondolas

print(min_gonlands([40, 50, 60, 70], 100))  # Output: 3
```

You are given two integer arrays:

g[] — greed factor of each child

s[] — size of available cookies

A child i is satisfied if they receive a cookie with size ≥ g[i].

Each child can receive at most one cookie.

Each cookie can be given to only one child.


Goal:

Return the maximum number of satisfied children.

Input:
g = [1, 2, 3]

s = [1, 1]

Output:

1



```python
def findContentChildren(g, s):
    g.sort()
    s.sort()
    
    i = 0
    j = 0
    satisfied = 0
    
    while i < len(g) and j < len(s):
        if s[j] >= g[i]:
            satisfied += 1
            i += 1
            j += 1
        else:
            j += 1
    
    return satisfied


print(findContentChildren([1, 2, 3], [1, 1]))
```
