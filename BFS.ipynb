### BFS: Breath First Tree LeetCode Questions

```python
## BFS Implementation

from collections import deque
def bfs(graph,start):
    queue=deque[start]
    visited=set()
    visited.add(start)
    
    while queue:
        node=queue.popleft()
        print(node,end=' ')
        for nei in graph[node]:
            if nei not in visited:
                visited.add(nei)
                queue.append(nei)
```
