```python
from typing import List

class Solution:
    def areSimilar(self, mat: List[List[int]], k: int) -> bool:
        roated_mat=copy.deepcopy(mat)
        for i in range(k):
            for j in range(len(mat)):
                if j%2==0:
                    first = roated_mat[j].pop(0)
                    roated_mat[j].append(first)
                else:
                    last=roated_mat[j].pop()
                    roated_mat[j].insert(0,last)
        return roated_mat ==mat
    
    
s=Solution()
mat = [[1,2,1,2],[5,5,5,5],[6,3,6,3]]
k = 2

print(s.areSimilar(mat,k))
```

```python
class Solution:
    def findMissingAndRepeatedValues(self, grid: List[List[int]]) -> List[int]:
        flatten_grid=[i for row in grid for i in row]
        n=len(grid)
        res=[]
        for i in range(1,n**2+1):
            if i not in flatten_grid:
                res.append(i)
        seen=set()
        for i in flatten_grid:
            if i in seen :
                res.append(i)
            else:
                seen.add(i)
        return res

s=Solution()
grid = grid = [[9,1,7],[8,9,2],[3,4,6]]

print(s.findMissingAndRepeatedValues(grid))
```
